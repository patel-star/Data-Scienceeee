```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('placement.csv')
df.shape
```

```python
df.sample(5)
```

```python
plt.figure(figsize = (16,5))
plt.subplot(1,2,1)
sns.histplot(df['cgpa'],kde='True')

plt.show()
# this is normally distributed so we can apply the trimming and capping:
```

```python
plt.subplot(1,2,1)
sns.histplot(df['placement_exam_marks'],kde='True')

plt.show()
# here the placement_exam_marks is not normally distributed so trimming and capping wont work:
```

```python
df.placement_exam_marks.skew()
# 0.83- right skewed moderate
```

```python
print('Mean value of cgpa:',df.cgpa.mean())
print('Std value of cgpa:',df.cgpa.std())
print('Min value of cgpa:',df.cgpa.min())
print('Max value of cgpa:',df.cgpa.max())
```

```python
# Now finding boundries values:
print('Highest allowed', df.cgpa.mean() + 3*df.cgpa.std())
print('Lowest allowed', df.cgpa.mean() - 3*df.cgpa.std())
```

```python
# Finding outliers (in cgpa):
df[(df.cgpa>8.8) | (df.cgpa<5.11)]
```

# Trimming:

```python
# approach-1:
new_df = df[(df['cgpa'] < 8.8) & (df['cgpa'] > 5.11)]
new_df
```

```python
# approach-2: calculating tha Zscore

df['cgpa_zscore'] = (df.cgpa - df.cgpa.mean())/df.cgpa.std()
df.head()
```

```python
df[df.cgpa_zscore > 3]
```

```python
df[df.cgpa_zscore < -3]
```

```python
df[(df['cgpa_zscore'] > 3) | (df['cgpa_zscore'] < -3)]
```

```python
# now again trimming based on zscore:
new_df = df[(df.cgpa_zscore<3) & (df.cgpa_zscore> -3)]
new_df
```

# Capping:

```python
# other outliers removal method is Capping:
```

```python
upper_limit = df.cgpa.mean() + 3*df.cgpa.std()
lower_limit = df.cgpa.mean() - 3*df.cgpa.std()
lower_limit
```

```python
df.cgpa = np.where(
    df.cgpa > upper_limit,
    upper_limit,
    np.where(
        df.cgpa < lower_limit,
        lower_limit,
        df.cgpa
    )
)
```
