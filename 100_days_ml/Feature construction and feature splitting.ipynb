```python
import numpy as np
import pandas as pd

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
import seaborn as sns
```

```python
df = pd.read_csv('Titanic.csv')[['Age','Pclass','SibSp','Parch','Survived']]
df.head()
```

```python
df.dropna(inplace = True)
df.head()
```

```python
x = df.iloc[:,0:4]
y = df.iloc[:,-1]
x.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),x,y,scoring='accuracy',cv=20))
```

```python
# Applying Feature Construction:
```

```python
x['Family_size'] = x.SibSp + x.Parch +1
x.head()
```

```python
def myfunc(num):
    if num ==1:      #alone
        return 0
    elif num > 1 and num <= 4:   #small famly
        return 1
    else:               #large family
        return 2

myfunc(4)
```

```python
x['Family_Type'] = x.Family_size.apply(myfunc)
x.head()
```

```python
x.drop(columns=['SibSp','Parch','Family_size'], inplace=True)
x.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),x,y,scoring='accuracy',cv=20))

# it increased from 63% to 73%:
```

```python
df = pd.read_csv('Titanic.csv')
df.head()
```

```python
# here the name have title like mr,mrs which can also reduce the accuracy:
df.Name
```

```python
df['Title'] = df.Name.str.split(', ', expand = True)[0]
df.Title
```

```python
df['Title'] = df.Name.str.split(', ', expand = True)[1]
df.Title
```

```python
df['Title'] = df.Name.str.split(', ', expand = True)[1].str.split('.', expand=True)[0]
df[['Title','Name']]
```

```python
# we can create married and unmaried now with mr, miss, mrs title:
```

```python
(df.groupby('Title')['Survived']).mean().sort_values(ascending=False)
# we can also see Mrs, Miss ,etc have high survival rate:
```

```python
df['Is_married'] = 0
df.loc[df.Title == 'Mrs', 'Is_married'] = 1
```
